## Goal: Merge all six months parquet files into one before starting with the cleaning process.

### Importing necessary libraries and merging all 6 into 1 using loop.

```python
import pandas as pd
import glob

# Get first 6 parquet files
files = sorted(glob.glob("yellow_tripdata_2025-*.parquet"))[:6]

# Read and combine
df = pd.concat(
    [pd.read_parquet(file) for file in files],
    ignore_index=True
)
```

#### After merging checking info of the merged dataset

```python
df.info()
```

This dataset has 24083384 rows and 20 columns with following datatypes: datetime64[us](2), float64(13), int32(3), int64(1), object(1)
and 3.3+ GB memory usage.

```python
df.to_parquet('yellow_taxi_6_months.parquet')
```
