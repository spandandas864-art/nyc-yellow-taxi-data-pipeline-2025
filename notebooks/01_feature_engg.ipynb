# Creating new features

#### fare_per_mile

```python
#fare per mile
pandas_df['fare_per_mile'] = pandas_df['fare_amount'] / pandas_df['trip_distance']
```

#### tip_percentage

```python
#tip pct(%)
pandas_df['tip_pct(%)'] = pandas_df['tip_amount'] / pandas_df['fare_amount']
```

#### trip_duration_in_minutes

```python
# Calculate trip duration in minutes
pandas_df['trip_duration_minutes'] = (pandas_df['tpep_dropoff_datetime'] - pandas_df['tpep_pickup_datetime']).dt.total_seconds() / 60
```

#### day_of_week, hour_of_the_day, month

```python
# Extract day of the week (Monday=0, Sunday=6)
pandas_df['pickup_day_of_week'] = pandas_df['tpep_pickup_datetime'].dt.dayofweek

# Extract hour of the day
pandas_df['pickup_hour'] = pandas_df['tpep_pickup_datetime'].dt.hour

# Extract month
pandas_df['pickup_month'] = pandas_df['tpep_pickup_datetime'].dt.month
```

#### is_weekend flag

```python
# Create a 'is_weekend' flag (Saturday=5, Sunday=6)
pandas_df['is_weekend'] = pandas_df['pickup_day_of_week'].isin([5, 6]).astype(int)
```
